## Effect of `max_iter` on model performance and convergence

To study the impact of the maximum number of iterations on model performance, I trained a logistic regression classifier using seven different values of `max_iter`. The tested values and corresponding accuracies (computed using the `score()` method) are summarized below:

- `max_iter = 7` → accuracy ≈ 0.806  
- `max_iter = 100` → accuracy ≈ 0.815  
- `max_iter = 500` → accuracy ≈ 0.840  
- `max_iter = 1000` → accuracy ≈ 0.842  
- `max_iter = 2000` → accuracy ≈ 0.842  
- `max_iter = 5000` → accuracy ≈ 0.842  
- `max_iter = 10000` → accuracy ≈ 0.842  

For small values of `max_iter`, the model fails to converge, as indicated by convergence warnings, and the resulting accuracy is relatively low. As `max_iter` increases, the optimization process is allowed to run longer, leading to improved convergence and higher accuracy.

However, the results show that increasing `max_iter` improves model accuracy only up to a point. Once convergence is reached (around `max_iter = 1000` in this case), further increases in the number of iterations do not lead to meaningful gains in accuracy. This indicates that the model has already reached its optimal solution.

## Computational cost considerations

Initially, I expected that increasing `max_iter` would significantly increase the computational cost, as measured by training time. However, in practice, the observed training time remained very similar across different values of `max_iter` (approximately 0–1 seconds).

This behavior can be explained by the fact that, in `LogisticRegression`, `max_iter` represents an upper bound rather than a fixed number of iterations. The optimization algorithm stops early once the convergence criterion is satisfied. As a result, even if `max_iter` is set to a large value (e.g. 3000 or 10000), the model may stop after around 1000 iterations if convergence is reached earlier.

Therefore, increasing `max_iter` increases the potential computational cost, but it does not necessarily increase the actual training time when the model converges early.

### Homework2

*Rescale all the features which are not dummies and train the logistic regression classifier with the default number of iterations. Do you get a warning about non-convergence now?*

`df.head()` is used to display the first five rows of the dataset in order to identify dummy variables. From the observed values, only `intplan` and `dataplan` are binary, while the remaining features are not dummies.


```python
df.head()
```

```python
non_dummy_cols = [
    'aclength', 'ommin', 'omcall', 'otmin', 'otcall',
    'ngmin', 'ngcall', 'imin', 'icall', 'cuscall'
]
```

Next, all non-dummy features were rescaled using `StandardScaler` in order to bring them to a comparable scale and improve the convergence behavior of the logistic regression model. The dummy variables were left unchanged to preserve their binary interpretation.




```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X = df.drop(columns='churn')
X_scaled = X.copy()

X_scaled[non_dummy_cols] = scaler.fit_transform(X[non_dummy_cols])
```

Finally, the logistic regression classifier was trained on the rescaled dataset using the default number of iterations.

```python
from sklearn.linear_model import LogisticRegression

y = df['churn']

clf = LogisticRegression()  # default max_iter = 100
clf.fit(X_scaled, y)
clf.score(X_scaled, y)
```

After rescaling all non-dummy features, the logistic regression classifier was trained using the default number of iterations (`max_iter = 100`). The model converged successfully without producing any warning about non-convergence, and the resulting accuracy was approximately 0.842. This confirms that feature scaling effectively improves the convergence behavior of the optimization algorithm and removes the need to increase the maximum number of iterations.
